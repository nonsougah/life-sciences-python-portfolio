# Introductory DNA Sequence Analysis with Biopython

**Portfolio demonstration — uses an illustrative sequence string, not a clinical or organism-identification result.**

## Objectives
- Represent a DNA sequence using Biopython.
- Calculate sequence length and GC percentage.
- Transcribe DNA to RNA and translate a coding sequence.
- Explain why sequence context and reference databases matter.

```python
from Bio.Seq import Seq
from Bio.SeqUtils import gc_fraction

dna = Seq("ATGGCCATTGTAATGGGCCGCTGAAAGGGTGCCCGATAG")
print("DNA:", dna)
print("Length:", len(dna), "nucleotides")
print("GC content:", round(gc_fraction(dna) * 100, 2), "%")
```

```python
rna = dna.transcribe()
protein = dna.translate(to_stop=True)

print("RNA:", rna)
print("Translated peptide (to first stop):", protein)
```

```python
# Simple sequence composition
from collections import Counter

counts = Counter(str(dna))
composition = {base: counts.get(base, 0) for base in "ACGT"}
composition
```

## Interpretation and limitations

This notebook demonstrates basic sequence operations only. Translation assumes the input is in the correct reading frame and represents a coding sequence under the standard genetic code. Real genomic analysis requires validated reference sequences, metadata, quality control, appropriate alignment or variant-calling methods, and careful interpretation.

No diagnosis, pathogenicity claim, organism identification, or clinical conclusion should be made from this illustrative sequence.
